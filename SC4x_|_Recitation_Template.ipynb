# Segment 4: Working with the Database

## Step 1: Set up the full database

>The code below will set up the full database with 5 tables imported from URLs and specify details about the file format and the data type for each column.

```python
# Create the full database

db_full = Database(f"""

    CREATE TABLE Locations AS
      SELECT *
      FROM read_csv('https://mit-sc4x-sharing.sfo3.cdn.digitaloceanspaces.com/python-sql-recitation/sc4x_wk2_recitation2_locations.csv',
        delim = ',',
        header = true,
        types = {{
          'Location_ID': 'INTEGER PRIMARY KEY',
          'Location_Territory': 'VARCHAR',
          'Location_Country': 'VARCHAR'
        }});

    CREATE TABLE Stores AS
      SELECT *
      FROM read_csv('https://mit-sc4x-sharing.sfo3.cdn.digitaloceanspaces.com/python-sql-recitation/sc4x_wk2_recitation2_stores.csv',
        delim = ',',
        header = true,
        types = {{
          'Store_ID': 'INTEGER PRIMARY KEY',
          'Store_Name': 'VARCHAR',
          'Store_Phone': 'VARCHAR',
          'Store_Manager': 'VARCHAR',
          'Location_ID': 'INTEGER REFERENCES Locations(Location_ID)'
        }});

      CREATE TABLE Warehouses AS
        SELECT *
        FROM read_csv('https://mit-sc4x-sharing.sfo3.cdn.digitaloceanspaces.com/python-sql-recitation/sc4x_wk2_recitation2_warehouses.csv',
          delim = ',',
          header = true,
          types = {{
            'Warehouse_ID': 'INTEGER PRIMARY KEY',
            'Warehouse_Name': 'VARCHAR',
            'Warehouse_Manager': 'VARCHAR',
            'Warehouse_Phone': 'VARCHAR',
            'Location_ID': 'INTEGER REFERENCES Locations(Location_ID)'
          }});

      CREATE TABLE Shipments AS
        SELECT *
        FROM read_csv('https://mit-sc4x-sharing.sfo3.cdn.digitaloceanspaces.com/python-sql-recitation/sc4x_wk2_recitation2_shipments.csv',
          delim = ',',
          header = true,
          types = {{
            'Shipment_ID': 'INTEGER PRIMARY KEY',
            'Warehouse_ID': 'INTEGER REFERENCES Warehouses(Warehouse_ID)',
            'Store_ID': 'INTEGER REFERENCES Stores(Store_ID)',
            'Transportation_Cost': 'DECIMAL(6,2)',
            'Total_Product_Value': 'DECIMAL(16,2)'
          }});

""")
```

```python
db_full.show_info()
```

## Step 2: Working with the Database

### Example 1: Basic Query

```python
# Example 1
# Warning: 'SELECT *' will pull all of the columns in the table, so use it cautiously for large tables.
pprint(db_full.query("""

    SELECT * FROM Warehouses;

"""))
```

### Example 2: Single Column, Adding `WHERE`

```python
# Example 2
pprint(db_full.query("""

    SELECT Warehouse_Name
    FROM Warehouses
    WHERE Warehouse_ID = 3;

"""))
```

### Example 3: Implicit Join

```python
# Example 3
pprint(db_full.query("""

    SELECT Locations.Location_Territory
    FROM Locations, Stores
    WHERE Stores.Store_Manager = 'Vanessa'
      AND
      Locations.Location_ID = Stores.Location_ID;

"""))
```

### Example 4: `JOIN` Syntax

```python
# Example 4
pprint(db_full.query("""

    SELECT Locations.Location_Territory
    FROM Stores
      JOIN Locations ON Locations.Location_ID = Stores.Location_ID
    WHERE Stores.Store_Manager = 'Vanessa';

"""))
```

### Example 5: Multiple Joins

```python
# Example 5
pprint(db_full.query("""

    SELECT Shipments.Shipment_ID, Shipments.Transportation_Cost
    FROM Shipments, Warehouses, Stores
    WHERE
      Warehouses.Warehouse_Name = 'The_Big_Red_Barn'
      AND
      Stores.Store_Manager = 'Victoria'
      AND
      Stores.Store_ID = Shipments.Store_ID
      AND
      Warehouses.Warehouse_ID = Shipments.Warehouse_ID;

"""))
```

### Example 6: `SUM` Function

```python
# Example 6
pprint(db_full.query("""

    SELECT SUM(Shipments.Transportation_Cost)
    FROM Shipments, Warehouses, Stores
    WHERE
      Warehouses.Warehouse_Name = 'The_Big_Red_Barn'
      AND
      Stores.Store_Manager = 'Victoria'
      AND
      Stores.Store_ID = Shipments.Store_ID
      AND
      Warehouses.Warehouse_ID = Shipments.Warehouse_ID;

"""))
```

### Example 7: `COUNT` Function

```python
# Example 7
pprint(db_full.query("""

    SELECT COUNT(Shipments.Transportation_Cost)
    FROM Shipments, Warehouses, Stores
    WHERE
      Warehouses.Warehouse_Name = 'The_Big_Red_Barn'
      AND
      Stores.Store_Manager = 'Victoria'
      AND
      Stores.Store_ID = Shipments.Store_ID
      AND
      Warehouses.Warehouse_ID = Shipments.Warehouse_ID;

"""))
```
